```python
import os
import numpy as np
import pandas as pd
from transformers import AutoTokenizer
from transformers import logging as hf_logging
hf_logging.set_verbosity_error()

data_dir = r"C:\Users\X\Desktop\X\ALTA\alta2026_public_data"
model_name = "roberta-base"

train_df = pd.read_csv(os.path.join(data_dir, "train.csv"))
valid_df = pd.read_csv(os.path.join(data_dir, "valid.csv"))

texts = pd.concat([train_df["text"], valid_df["text"]]).astype(str).tolist()

tokenizer = AutoTokenizer.from_pretrained(model_name)

lengths = []

for i in range(0, len(texts), 1000):
    batch = texts[i:i + 1000]
    encoded = tokenizer(
        batch,
        add_special_tokens=True,
        truncation=False
    )
    lengths.extend(len(ids) for ids in encoded["input_ids"])

lengths = np.array(lengths)

print(f"n={len(lengths)}")
print(f"mean={lengths.mean():.2f} std={lengths.std():.2f} min={lengths.min()} max={lengths.max()}")

for p in [50, 75, 90, 95, 97, 98, 99, 99.5, 100]:
    print(f"p{p}={np.percentile(lengths, p):.1f}")

for cand in [32, 48, 64, 96, 128, 160, 192, 256]:
    coverage = (lengths <= cand).mean() * 100
    print(f"max_length={cand} coverage={coverage:.2f}%")
```
